**Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data.** The program should:
Convert text to lowercase. (Also display how many uppercase?)
Remove punctuation.        (Also display how many different types of punctuations?)
Remove numbers.            (Also display how many numbers?)
Remove extra whitespace. (How much extra whitespace is present?)
Display the cleaned text.   


```python
from google.colab import files
uploaded = files.upload()
```

```python
import re
import string
# Read the text file
with open("2.1_text_data.txt", "r", encoding="utf-8") as file:
  content=file.read()
  print(content)
```

```python
# 1. Convert text to lowercase
uppercase_count = sum(1 for char in content if char.isupper())
text = content.lower()

print("\nNumber of uppercase letters:", uppercase_count)
```

```python
# 2. Remove punctuation
punctuations = set(char for char in text if char in string.punctuation)
punctuation_count = sum(1 for char in text if char in string.punctuation)

text = text.translate(str.maketrans("", "", string.punctuation))

print("Total punctuation characters:", punctuation_count)
print("Different types of punctuation:", len(punctuations))
print("Punctuation types:", punctuations)
```

```python
# 3. Remove numbers
number_count = sum(1 for char in text if char.isdigit())

text = re.sub(r'\d+', '', text)

print("Number of numbers:", number_count)
```

```python
# 4. Remove extra whitespace
extra_whitespace = len(re.findall(r'\s{2,}', text))

text = re.sub(r'\s+', ' ', text).strip()

print("Extra whitespace occurrences:", extra_whitespace)
```

```python
# 5. Display cleaned text
print("\nCleaned Text:")
print(text)
```

**Experiment 2.2:** Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)

```python
from google.colab import files
uploaded = files.upload()
```

```python
#using NLTK
import nltk

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Read the file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Word Tokenization
words = nltk.word_tokenize(text)

# Sentence Tokenization
sentences = nltk.sent_tokenize(text)

print("WORD TOKENS:")
print(words)

print("\nSENTENCE TOKENS:")
for i, sentence in enumerate(sentences, 1):
    print(i, ".", sentence)
```

```python
#using spaCy
import spacy

# Create a blank English NLP pipeline
nlp = spacy.blank("en")

# Add sentence segmentation
nlp.add_pipe("sentencizer")

# Read the file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process the text
doc = nlp(text)

# Word Tokenization
words = [token.text for token in doc]

# Sentence Tokenization
sentences = [sentence.text for sentence in doc.sents]

print("WORD TOKENS:")
print(words)

print("\nSENTENCE TOKENS:")
for i, sentence in enumerate(sentences, 1):
    print(i, ".", sentence)
```

```python
#without NLTK and spaCy
import re

# Read the file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Sentence Tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

# Word Tokenization
words = re.findall(r'\b\w+\b', text)

print("WORD TOKENS:")
print(words)

print("\nSENTENCE TOKENS:")
for i, sentence in enumerate(sentences, 1):
    print(i, ".", sentence)
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
from google.colab import files

uploaded = files.upload()
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Read the text file
with open("2.3_clean_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Tokenize the text
tokens = word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words('english'))

# Find stop words
found_stop_words = [
    word for word in tokens
    if word.lower() in stop_words
]

# Remove stop words
filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words
]

# Display original tokens
print("ORIGINAL TOKENS:")
print(tokens)

# Display stop words separately
print("\nSTOP WORDS:")
print(found_stop_words)

# Display filtered tokens
print("\nFILTERED TOKENS:")
print(filtered_tokens)
```
