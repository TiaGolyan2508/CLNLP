## Experiment 5.1: Implement subword-level tokenization using a BPE-based tokenizer.

- **Aim:** Tokenize the given text into subword units and display each token with its corresponding token ID.
- **Part A:** Use a pretrained BPE tokenizer.
- **Part B:** Train a BPE tokenizer without using a pretrained model.

```python
from pathlib import Path

input_filename = "input_sub_word_data (1).txt"
possible_paths = [
    Path(input_filename),
    Path("/content") / input_filename,
    Path("/content/sample_data") / input_filename
]

input_path = next((p for p in possible_paths if p.exists()), None)

if input_path is None:
    try:
        from google.colab import files
        uploaded = files.upload()
        input_path = Path(next(iter(uploaded)))
    except ImportError:
        raise FileNotFoundError(
            "Place 'input_sub_word_data (1).txt' in the notebook folder and run again."
        )

text = input_path.read_text(encoding="utf-8")
print("Input file:", input_path)
print("Number of characters:", len(text))
print("\nFirst 500 characters:\n")
print(text[:500])
```

### **5.1.A** BPE tokenization using a pretrained model

```python
from transformers import AutoTokenizer

# GPT-2 uses Byte Pair Encoding (BPE)
model = AutoTokenizer.from_pretrained("gpt2")

tokens = model.tokenize(text)
token_ids = model.convert_tokens_to_ids(tokens)

print("BPE TOKENIZATION USING PRETRAINED GPT-2")
print("-" * 65)
print(f'{"Token":<30} {"Token ID"}')
print("-" * 65)

for token, token_id in zip(tokens, token_ids):
    print(f"{token!r:<30} {token_id}")

print("\nTotal tokens:", len(tokens))
```

### **5.1.B** BPE tokenization without using a pretrained model

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers, normalizers

# Create a new BPE tokenizer and learn its vocabulary from the input text
tokenizer_bpe_scratch = Tokenizer(models.BPE(unk_token="[UNK]"))
tokenizer_bpe_scratch.normalizer = normalizers.Lowercase()
tokenizer_bpe_scratch.pre_tokenizer = pre_tokenizers.Whitespace()

bpe_trainer = trainers.BpeTrainer(
    vocab_size=500,
    min_frequency=2,
    special_tokens=["[UNK]", "[PAD]"]
)

tokenizer_bpe_scratch.train_from_iterator([text], trainer=bpe_trainer)
encoded_bpe = tokenizer_bpe_scratch.encode(text)

print("BPE TOKENIZATION WITHOUT PRETRAINED MODEL")
print("-" * 65)
print(f'{"Token":<30} {"Token ID"}')
print("-" * 65)

for token, token_id in zip(encoded_bpe.tokens, encoded_bpe.ids):
    print(f"{token!r:<30} {token_id}")

print("\nTotal tokens:", len(encoded_bpe.tokens))
print("Learned vocabulary size:", tokenizer_bpe_scratch.get_vocab_size())
```

## Experiment 5.2: Implement subword-level tokenization using a SentencePiece tokenizer.

- **Aim:** Tokenize the same input text into subword units and display each token with its token ID.
- **Part A:** Use a pretrained SentencePiece model.
- **Part B:** Train a SentencePiece model without using a pretrained model.

### **5.2.A** SentencePiece tokenization using a pretrained model

The pretrained `t5-small` tokenizer is based on SentencePiece.

```python
from transformers import AutoTokenizer

tokenizer_sp_pretrained = AutoTokenizer.from_pretrained("t5-small")

tokens_sp = tokenizer_sp_pretrained.tokenize(text)
ids_sp = tokenizer_sp_pretrained.convert_tokens_to_ids(tokens_sp)

print("SENTENCEPIECE TOKENIZATION USING PRETRAINED T5")
print("-" * 65)
print(f'{"Token":<30} {"Token ID"}')
print("-" * 65)

for token, token_id in zip(tokens_sp, ids_sp):
    print(f"{token!r:<30} {token_id}")

print("\nTotal tokens:", len(tokens_sp))
```

### **5.2.B** SentencePiece tokenization without using a pretrained model

```python
import sentencepiece as spm

# Train a new SentencePiece model on the input corpus.
# Unigram is used here as the SentencePiece model type.
spm.SentencePieceTrainer.train(
    sentence_iterator=iter(line for line in text.splitlines() if line.strip()),
    model_prefix="my_sentencepiece",
    vocab_size=200,
    model_type="unigram",
    character_coverage=1.0,
    hard_vocab_limit=False
)

sentencepiece_model = spm.SentencePieceProcessor(
    model_file="my_sentencepiece.model"
)

tokens_sp_scratch = sentencepiece_model.encode(text, out_type=str)
ids_sp_scratch = sentencepiece_model.encode(text, out_type=int)

print("SENTENCEPIECE TOKENIZATION WITHOUT PRETRAINED MODEL")
print("-" * 65)
print(f'{"Token":<30} {"Token ID"}')
print("-" * 65)

for token, token_id in zip(tokens_sp_scratch, ids_sp_scratch):
    print(f"{token!r:<30} {token_id}")

print("\nTotal tokens:", len(tokens_sp_scratch))
print("Learned vocabulary size:", sentencepiece_model.get_piece_size())
```

## Experiment 5.3: Perform Part-of-Speech (POS) tagging using spaCy and NLTK.

**Given sentence:** `The young student is reading an interesting book in the library.`

- **Aim:** Display each unique token, its POS tag, and the description of the POS tag.

### **5.3.A** POS tagging using spaCy

```python
import spacy

# Run this once if the English model is not installed:
# !python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")
sentence = "The young student is reading an interesting book in the library."
doc = nlp(sentence)

unique_pos = {}

for token in doc:
    if not token.is_space and token.text not in unique_pos:
        unique_pos[token.text] = (token.pos_, spacy.explain(token.pos_))

print(f'{"Token":<18} {"POS Tag":<12} {"Description"}')
print("-" * 60)

for token, (pos_tag, description) in unique_pos.items():
    print(f"{token:<18} {pos_tag:<12} {description}")
```

### **5.3.B** POS tagging using NLTK

```python
import nltk

# Download resources if they are not already installed
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

sentence = "The young student is reading an interesting book in the library."
tokens = nltk.word_tokenize(sentence)
tagged_tokens = nltk.pos_tag(tokens)

# Keep the first POS tag for each unique token
unique_nltk_pos = {}
for token, tag in tagged_tokens:
    if token not in unique_nltk_pos:
        unique_nltk_pos[token] = tag

tag_descriptions = {
    "DT": "Determiner",
    "JJ": "Adjective",
    "JJR": "Comparative adjective",
    "JJS": "Superlative adjective",
    "NN": "Singular noun",
    "NNS": "Plural noun",
    "NNP": "Proper noun",
    "NNPS": "Plural proper noun",
    "VB": "Base form verb",
    "VBD": "Past tense verb",
    "VBG": "Gerund or present participle",
    "VBN": "Past participle",
    "VBP": "Present tense verb",
    "VBZ": "Third-person singular present verb",
    "IN": "Preposition or subordinating conjunction",
    "RB": "Adverb",
    "PRP": "Personal pronoun",
    "CC": "Coordinating conjunction",
    ".": "Sentence-ending punctuation",
    ",": "Comma"
}

print(f'{"Token":<18} {"POS Tag":<12} {"Description"}')
print("-" * 60)

for token, tag in unique_nltk_pos.items():
    description = tag_descriptions.get(tag, "Other")
    print(f"{token:<18} {tag:<12} {description}")
```

## Experiment 5.4: Perform POS tagging and calculate token frequency using spaCy.

- **Aim:** For the same input text used in Experiment 5.1, display each unique token, its POS tag, POS description, and frequency.

```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

# Count each token exactly as it appears (case-sensitive; punctuation included)
frequency = Counter(token.text for token in doc if not token.is_space)

# Store POS tag and description once for each unique token
pos_information = {}
for token in doc:
    if not token.is_space and token.text not in pos_information:
        pos_information[token.text] = (
            token.pos_,
            spacy.explain(token.pos_)
        )

print(f'{"Token":<25} {"POS Tag":<12} {"Description":<25} {"Frequency"}')
print("-" * 90)

for token, (pos_tag, description) in pos_information.items():
    print(
        f"{token:<25} {pos_tag:<12} {description:<25} "
        f"{frequency[token]}"
    )
```

## Experiment 5.5: Train a Word2Vec CBOW model.

- **Aim:** Train a Continuous Bag of Words (CBOW) model using the given corpus, display the vector representation of a selected word, and find the three most similar words.
- **Input:** Create `corpus.txt` using the sentences below.

```python
corpus = [
    "Natural language processing is interesting",
    "Natural language processing is useful",
    "I love machine learning",
    "Machine learning is useful",
    "Deep learning is a part of machine learning",
    "Artificial intelligence is changing the world",
    "Natural language processing uses machine learning",
    "Deep learning and artificial intelligence are related",
    "I study natural language processing",
    "Machine learning helps artificial intelligence"
]

with open("corpus.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(corpus))

print("corpus.txt created successfully.")
```

```python

!pip install gensim

import re
from gensim.models import Word2Vec

# Read and preprocess the corpus
with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        re.findall(r"\b\w+\b", line.lower())
        for line in file if line.strip()
    ]

# sg=0 selects CBOW
cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=0,
    seed=42,
    epochs=100
)

selected_word = "machine"

print("Selected word:", selected_word)
print("\nVector representation:")
print(cbow_model.wv[selected_word])

print("\nThree most similar words:")
for word, score in cbow_model.wv.most_similar(selected_word, topn=3):
    print(f"{word}: {score:.4f}")
```

## Experiment 5.6: Train a Skip-Gram Word2Vec model.

- **Aim:** Train a Skip-Gram model using `corpus.txt`, retrieve the vector representation of `language`, and display the five most similar words.

```python
import re
from gensim.models import Word2Vec

with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        re.findall(r"\b\w+\b", line.lower())
        for line in file if line.strip()
    ]

# sg=1 selects Skip-Gram
skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=100
)

selected_word = "language"

print("Selected word:", selected_word)
print("\nVector representation:")
print(skipgram_model.wv[selected_word])

print("\nFive most similar words:")
for word, score in skipgram_model.wv.most_similar(selected_word, topn=5):
    print(f"{word}: {score:.4f}")
```

## Experiment 5.7: Calculate cosine similarity between selected word pairs using Word2Vec.

- **Aim:** Train a Word2Vec model on `corpus.txt` and calculate cosine similarity for:
  1. `machine` and `learning`
  2. `natural` and `language`
  3. `deep` and `learning`

```python
import re
from gensim.models import Word2Vec

with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        re.findall(r"\b\w+\b", line.lower())
        for line in file if line.strip()
    ]

similarity_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1,
    sg=0,
    seed=42,
    epochs=100
)

word_pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

print("COSINE SIMILARITY")
print("-" * 45)

for word1, word2 in word_pairs:
    score = similarity_model.wv.similarity(word1, word2)
    print(f"{word1} and {word2}: {score:.4f}")
```
